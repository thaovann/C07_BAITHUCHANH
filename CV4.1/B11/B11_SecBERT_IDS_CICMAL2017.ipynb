```python
# ============================================================
# TINYBERT - FIXED + FAST (< 10 PHÚT)
# ============================================================

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import seaborn as sns
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.utils import resample
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    f1_score,
    recall_score
)

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer
)

from datasets import Dataset

# ============================================================
# 1. DEVICE
# ============================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
print("⚡ GPU available!" if torch.cuda.is_available() else "⚠️  CPU mode — sẽ chậm hơn, ~6-9 phút")

# ============================================================
# 2. LOAD DATA
# ============================================================
DATA_PATH = Path.cwd().parent / "Dataset" / "category_CICMAL2017_dataset.csv"
df = pd.read_csv(DATA_PATH, low_memory=False)

print("Original shape:", df.shape)

# ============================================================
# 3. DROP META COLUMNS
# ============================================================
drop_cols = [
    "Flow ID",
    "Source IP",
    "Source Port",
    "Destination IP",
    "Destination Port",
    "Protocol",
    "Timestamp",
    "Cat_Label",  # label multi-class
    "Binary Label",  # label binary
]

# ============================================================
# 4. BALANCED SAMPLING — FAST MODE (1500 + 1500 = 3000)
# ============================================================
SAMPLES_PER_CLASS = 1500   # ⚡ Giảm từ 2500 → 1500

df_normal = df[df["Binary Label"] == 0]
df_attack  = df[df["Binary Label"] == 1]

df_normal_sampled = resample(
    df_normal,
    n_samples=min(SAMPLES_PER_CLASS, len(df_normal)),
    random_state=42,
    replace=len(df_normal) < SAMPLES_PER_CLASS
)
df_attack_sampled = resample(
    df_attack,
    n_samples=min(SAMPLES_PER_CLASS, len(df_attack)),
    random_state=42,
    replace=len(df_attack) < SAMPLES_PER_CLASS
)

df_balanced = pd.concat([df_normal_sampled, df_attack_sampled]).sample(
    frac=1, random_state=42
).reset_index(drop=True)
print(df_balanced.columns)
X = df_balanced.drop(columns=drop_cols)
y = df_balanced["Binary Label"].values

X = X.replace([np.inf, -np.inf], np.nan).fillna(0)

print(f"Balanced dataset: {X.shape}  |  Normal={(y==0).sum()}  Attack={(y==1).sum()}")

# ============================================================
# 5. CONVERT NUMERIC → TEXT
# ============================================================
def to_text(row):
    return " ".join([f"f{i}={v:.3f}" for i, v in enumerate(row)])

texts = np.array([to_text(r) for r in X.values])

# ============================================================
# 6. SPLIT
# ============================================================
train_texts, test_texts, train_labels, test_labels = train_test_split(
    texts,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print(f"Train: {len(train_texts)}  |  Test: {len(test_texts)}")

# ============================================================
# 7. MODEL (TINYBERT)
# ============================================================
MODEL_NAME = "huawei-noah/TinyBERT_General_4L_312D"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2
).to(device)

# ============================================================
# 8. TOKENIZE — max_length=64 (⚡ giảm từ 128)
# ============================================================
train_enc = tokenizer(list(train_texts), truncation=True, padding=True, max_length=64)
test_enc  = tokenizer(list(test_texts),  truncation=True, padding=True, max_length=64)

train_dataset = Dataset.from_dict({
    "input_ids":      train_enc["input_ids"],
    "attention_mask": train_enc["attention_mask"],
    "labels":         train_labels.tolist()
})

test_dataset = Dataset.from_dict({
    "input_ids":      test_enc["input_ids"],
    "attention_mask": test_enc["attention_mask"],
    "labels":         test_labels.tolist()
})

# ============================================================
# 9. CLASS WEIGHTS FOR WEIGHTED LOSS
# ============================================================
n_normal = (train_labels == 0).sum()
n_attack = (train_labels == 1).sum()
total    = len(train_labels)

w_normal = total / (2 * n_normal)
w_attack = total / (2 * n_attack)

class_weights = torch.tensor([w_normal, w_attack], dtype=torch.float).to(device)
print(f"Class weights — Normal: {w_normal:.4f}  |  Attack: {w_attack:.4f}")

# ============================================================
# 10. CUSTOM TRAINER WITH WEIGHTED CROSS-ENTROPY
# ============================================================
class WeightedTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels = inputs.pop("labels")
        outputs = model(**inputs)
        logits  = outputs.logits
        loss    = nn.CrossEntropyLoss(weight=class_weights)(logits, labels)
        return (loss, outputs) if return_outputs else loss

# ============================================================
# 11. MACRO F1 METRIC (sklearn — không cần cài thêm)
# ============================================================
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=1)
    return {
        "macro_f1":   float(f1_score(labels, preds, average="macro")),
        "rec_normal": float(recall_score(labels, preds, pos_label=0)),
        "rec_attack": float(recall_score(labels, preds, pos_label=1)),
    }

# ============================================================
# 12. TRAINING ARGUMENTS — FAST
# ============================================================
training_args = TrainingArguments(
    output_dir="./tinybert_fast_fixed",

    eval_strategy="epoch",
    save_strategy="no",

    learning_rate=2e-5,
    num_train_epochs=2,              # ⚡ 2 epoch (cân bằng tốc độ/accuracy)
    warmup_ratio=0.05,               # ⚡ warmup ngắn hơn

    per_device_train_batch_size=64,  # ⚡ tăng từ 32 → 64
    per_device_eval_batch_size=64,

    logging_steps=20,
    report_to="none",
    metric_for_best_model="macro_f1",

    fp16=torch.cuda.is_available()   # tự động dùng fp16 nếu có GPU
)

trainer = WeightedTrainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    compute_metrics=compute_metrics
)

print("\n🚀 Bắt đầu training...")
trainer.train()

# ============================================================
# 13. PREDICT
# ============================================================
pred    = trainer.predict(test_dataset)
logits  = pred.predictions
y_pred  = np.argmax(logits, axis=1)
y_proba = torch.softmax(torch.tensor(logits), dim=1).numpy()[:, 1]

# ============================================================
# 14. METRICS
# ============================================================
print("\n=== CLASSIFICATION REPORT ===")
print(classification_report(
    test_labels,
    y_pred,
    target_names=["Bình thường", "Tấn công"]
))

roc = roc_auc_score(test_labels, y_proba)
print(f"ROC-AUC: {roc:.4f}")

# ============================================================
# 15. CONFUSION MATRIX
# ============================================================
cm = confusion_matrix(test_labels, y_pred)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.heatmap(
    cm,
    annot=True, fmt="d", cmap="Blues",
    xticklabels=["Bình thường", "Tấn công"],
    yticklabels=["Bình thường", "Tấn công"],
    ax=axes[0]
)
axes[0].set_title("Ma trận nhầm lẫn - TinyBERT")
axes[0].set_xlabel("Dự đoán")
axes[0].set_ylabel("Thực tế")
```
