```python
# ============================================================
# DGA DETECTION USING TINYBERT
# ============================================================

import numpy as np
import torch
import seaborn as sns
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments
)

from datasets import Dataset

# =========================
# 1. DEVICE
# =========================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# =========================
# 2. LOAD DATA
# =========================
root = Path.cwd()
data = np.load(
    root / "Dataset" / "1000_dga_dataset_binary.npz",
    allow_pickle=True
)

X = data["X"]
y = data["y"]
char2idx = data["char2idx"].item()

print("X shape:", X.shape)
print("y shape:", y.shape)

# =========================
# 3. DECODE DOMAIN
# =========================
idx2char = {v: k for k, v in char2idx.items()}

domains = []

for row in X:
    domain = "".join(
        idx2char[idx]
        for idx in row
        if idx != 0
    )

    domains.append(domain)

print("\n5 domain đầu tiên:")

for i in range(5):
    print(domains[i], y[i])

# =========================
# 4. TRAIN TEST SPLIT
# =========================
train_texts, test_texts, train_labels, test_labels = train_test_split(
    domains,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("\nTrain:", len(train_texts))
print("Test:", len(test_texts))

# =========================
# 5. LOAD TINYBERT
# =========================
MODEL_NAME = "huawei-noah/TinyBERT_General_4L_312D"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    num_labels=2
).to(device)

# =========================
# 6. TOKENIZATION
# =========================
train_encodings = tokenizer(
    train_texts,
    truncation=True,
    padding=True,
    max_length=64
)

test_encodings = tokenizer(
    test_texts,
    truncation=True,
    padding=True,
    max_length=64
)

train_dataset = Dataset.from_dict({
    "input_ids": train_encodings["input_ids"],
    "attention_mask": train_encodings["attention_mask"],
    "labels": train_labels
})

test_dataset = Dataset.from_dict({
    "input_ids": test_encodings["input_ids"],
    "attention_mask": test_encodings["attention_mask"],
    "labels": test_labels
})

# =========================
# 7. METRICS
# =========================
def compute_metrics(eval_pred):
    preds = np.argmax(eval_pred.predictions, axis=-1)

    return {
        "accuracy": accuracy_score(test_labels, preds),
        "precision": precision_score(test_labels, preds),
        "recall": recall_score(test_labels, preds),
        "f1": f1_score(test_labels, preds)
    }

# =========================
# 8. TRAINING CONFIG
# =========================
training_args = TrainingArguments(
    output_dir="./tinybert_dga",

    eval_strategy="epoch",
    save_strategy="no",

    learning_rate=2e-5,
    num_train_epochs=2,

    per_device_train_batch_size=32,
    per_device_eval_batch_size=32,

    weight_decay=0.01,

    logging_steps=20,

    fp16=torch.cuda.is_available(),

    load_best_model_at_end=False,
    report_to="none"
)

# =========================
# 9. TRAINER
# =========================
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    compute_metrics=compute_metrics
)

# =========================
# 10. TRAIN
# =========================
trainer.train()

# =========================
# 11. PREDICT
# =========================
predictions = trainer.predict(test_dataset)

y_pred = np.argmax(
    predictions.predictions,
    axis=1
)

# =========================
# 12. CLASSIFICATION REPORT
# =========================
print("\nCLASSIFICATION REPORT")

print(
    classification_report(
        test_labels,
        y_pred,
        target_names=["Bình thường", "DGA"]
    )
)

# =========================
# 13. CONFUSION MATRIX
# =========================
cm = confusion_matrix(test_labels, y_pred)

plt.figure(figsize=(5, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Bình thường', 'DGA'],
    yticklabels=['Bình thường', 'DGA']
)

plt.title("Ma trận nhầm lẫn - TinyBERT")
plt.xlabel("Dự đoán")
plt.ylabel("Thực tế")

plt.tight_layout()
plt.show()
```
